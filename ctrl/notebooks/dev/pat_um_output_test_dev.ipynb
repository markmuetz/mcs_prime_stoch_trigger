```python
# /home/d02/mamue/projects/vn13.0_org_conv/src/atmosphere/convection/org_conv.F90
# IF (ALLOCATED(psif2)) THEN
#  org_temp_tendency(i,j,1) = psif2(i, j)
# END IF
# org_temp_tendency(i,j,2) = lcbase(i,j)
# org_temp_tendency(i,j,3) = lctop(i,j)
# org_temp_tendency(i,j,4) = p_layer_boundaries(i,j,lctop(i,j))
# org_temp_tendency(i,j,5) = p_layer_boundaries(i,j,lcbase(i,j)-1)
# org_temp_tendency(i,j,6) = lctop_temp
idx_to_name_map = {
    0: 'stoch_pattern',
    1: 'lcbase',
    2: 'lctop',
    3: 'pressure_cbase',
    4: 'pressure_ctop',
    5: 'lctop_temp',
}
units = {
    0: '-',
    1: '-',
    2: '-',
    3: 'hPa',
    4: 'hPa',
    5: 'K',
}

ds = xr.Dataset({
    idx_to_name_map[i]: da.isel(model_level_number=i).drop_vars("model_level_number").assign_attrs({
        'units': units[i]
    })
    for i in range(da.sizes["model_level_number"])
})
```

```python
tcwv = xr.DataArray.from_iris(pa000.extract_cube('m01s30i461')).rename('tcwv')
```

```python
tcwv
```

```python
import numpy as np

def calc_tcwv_prob(tcwv):
    p = 0.009066 * tcwv + 0.199061
    p = xr.where(p > 1, 1, p)
    return p
```

```python
tcwv_prob = calc_tcwv_prob(tcwv)
```

```python
tcwv_prob
```

```python
fig, ax = plt.subplots(subplot_kw={'projection': ccrs.PlateCarree()})
im = ax.pcolormesh(tcwv_prob.longitude, tcwv_prob.latitude, tcwv_prob[0])
plt.colorbar(im)
```

```python
fig, ax = plt.subplots(subplot_kw={'projection': ccrs.PlateCarree()})
im = ax.pcolormesh(tcwv_prob.longitude, tcwv_prob.latitude, tcwv_prob[1])
```

```python
import scipy
mu = 0.0
sigma = 0.26962024
z = (ds.stoch_pattern - mu) / sigma

pattern_prob = 0.5 * scipy.special.erfc(z / np.sqrt(2.0))
```

```python
fig, ax = plt.subplots(subplot_kw={'projection': ccrs.PlateCarree()})
im = ax.pcolormesh(tcwv_prob.longitude, tcwv_prob.latitude, pattern_prob[1] < tcwv_prob[1])
```

```python
fig, ax = plt.subplots(subplot_kw={'projection': ccrs.PlateCarree()})
im = ax.pcolormesh(tcwv_prob.longitude, tcwv_prob.latitude, pattern_prob[1] < 0.3978)
```
